```python
import pandas as pd
from sklearn.preprocessing import StandardScaler,LabelEncoder, OneHotEncoder
from sklearn.model_selection import train_test_split
import pickle
```

```python
data = pd.read_csv('Churn_Modelling.csv')
data.head()
```

```python
data=data.drop(['RowNumber','CustomerId','Surname'],axis=1,)
data
```

```python
## encode the gender column using label encoder

le_gender=LabelEncoder()
data['Gender']=le_gender.fit_transform(data['Gender'])
data.head()
```

```python
## One hot encoding for the 'Geography' column
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False)
geo_encoder = encoder.fit_transform(data[['Geography']])
feature_names = encoder.get_feature_names_out(['Geography'])

print(feature_names)
print(geo_encoder)
```

```python
geo_encoded_df = pd.DataFrame(geo_encoder, columns=feature_names)
geo_encoded_df
```

```python
data.head()
```

```python
## combine the encoded columns with the original dataframe

data=pd.concat([data.drop(columns=['Geography']), geo_encoded_df], axis=1)
data
```

```python
## divide the data into features (independent variables) and target variable(dependent variable)
X=data.drop('EstimatedSalary',axis=1)
y=data['EstimatedSalary']

## split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
X_train.shape
```

```python
X_train, x_val, y_train, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.2,
    random_state=0
)
```

```python
X_train.shape,x_val.shape,y_train.shape,y_val.shape
```

```python
X_train
```

```python
## Scale the features using StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
x_val = scaler.transform(x_val)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
## save the encoder and scalar
with open('onehot_encoder_geo_salary.pkl', 'wb') as f:
    pickle.dump(encoder, f)
with open('label_encoder_gender_salary.pkl', 'wb') as f:
    pickle.dump(le_gender, f)
with open('scaler_salary.pkl', 'wb') as f:
    pickle.dump(scaler, f)
```

## ANN Implementation

## ANN Architecture

1. BUILD   
   ↓  
   model = Sequential(...)  

2. COMPILE  
   ↓  
   model.compile(...)  
   
3. TRAIN   
   ↓  
   model.fit(...)  
  
4. PREDICT  
   ↓  
   model.predict(...)  

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime
```

## 1. BUILD

```python
## Build the ANN model
salary_model= Sequential()
salary_model.add(Dense(units=64, activation='relu', input_dim=12))
salary_model.add(Dense(units=32, activation='relu'))
salary_model.add(Dense(units=1,))
```

```python
salary_model.summary()
```

## 2. COMPILE

```python
## compile the model
salary_model.compile(optimizer='adam', loss="mae", metrics=['mae'])
```

### 3. TRAIN

```python
## set up TensorBoard callback
log_dir = "logs/fit/salaryregression_/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorboard_callback = TensorBoard(
    log_dir=log_dir,
    histogram_freq=1
)
```

```python
## set up EarlyStopping callback to prevent overfitting and stop training when the validation loss stops improving
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True
)
```

```python
model_history = salary_model.fit(
    X_train,
    y_train,
    epochs=500,
    validation_data=(x_val, y_val),
    callbacks=[tensorboard_callback, early_stopping]
)
```

```python
salary_model.save('salary_model.h5')
```

### 4. PREDICT

```python
# evaluate the model on the test set
test_loss, test_mae = salary_model.evaluate(X_test, y_test)
print(f"Test Loss: {test_loss}, Test MAE: {test_mae}")
```
