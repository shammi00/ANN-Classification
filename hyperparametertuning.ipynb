### Determining the optimal number of hidden layers and neurons for an
### Artificial Neural Network (ANN)

This can be challenging and often requires experimentation. However, there are some guidelines and method
that can help you in making an informed decision:

Start Simple: Begin with a simple architecture and gradually increase complexity if needed. 

Grid Search/Random Search: Use grid search or random search to try different architectures. 

Cross-Validation: Use cross-validation to evaluate the performance of different architectures. 

Heuristics and Rules of Thumb: Some heuristics and empirical rules can provide starting points, such as: 

o The number of neurons in the hidden layer should be between the size of the input layer and the
size of the output layer.  
 
o A common practice is to start with 1-2 hidden layers.

Grid Search/Random Search:
Decide which architectures to try.

Cross-Validation:
Decide how to evaluate those architectures reliably.

Grid Search = “Which settings should I use?”  
Cross-Validation = “How reliably does this setting perform?”

Grid Search = searches through the hyperparameter combinations

Cross-Validation = evaluates each combination more reliably

When they say "Grid Search" for ANN, they aren't necessarily talking about only GridSearchCV from scikit-learn.

For neural networks, you might use tools such as KerasTuner to search architectures/hyperparameters, while cross-validation is a separate evaluation technique.

So don't memorize:

"Grid Search = Cross Validation"

Instead remember:

Grid Search → SEARCH 🔍  
Cross-Validation → EVALUATE 📊

One important distinction

You can use cross-validation without Grid Search:

## cross_val_score(model, X, y, cv=5)

You're evaluating one particular model/settings.

And Grid Search normally looks like:

## GridSearchCV(model, param_grid, cv=5)

Here GridSearchCV means:

Grid Search + Cross-Validation

So what is the purpose of this 4th point?

The first three methods are more systematic:

1. Start Simple

Start with a simple architecture and increase complexity if needed.

2. Grid/Random Search

Automatically try different architectures/hyperparameters.

3. Cross-Validation

Evaluate those architectures more reliably.

4. Heuristics / Rules of Thumb

Use experience-based guidelines to decide where to start and what values are reasonable to try.

    "KerasClassifier is a tool that makes a Keras model behave like a scikit-learn estimator."

    What's a "scikit-learn estimator"?
    In scikit-learn, every model (like LogisticRegression, RandomForestClassifier, SVC, etc.) follows a common pattern — it's called an estimator. All estimators support the same basic methods:

    model.fit(X_train, y_train)      # train the model
    model.predict(X_test)             # make predictions
    model.score(X_test, y_test)       # evaluate accuracy

    Because every scikit-learn model follows this exact same pattern, tools like GridSearchCV or cross_val_score know exactly how to work with any of them — they just call .fit(), .predict(), etc.

    The problem with a Keras model
    A Keras model (like your ANN built with Sequential) does not follow that same pattern. Instead, it uses:

    model.fit(X_train, y_train, epochs=10, batch_size=32)
    model.predict(X_test)

    It looks similar, but Keras models don't fully match the strict scikit-learn "shape" that tools like GridSearchCV expect internally. So scikit-learn doesn't recognize a raw Keras model as something it can plug into its tools.

    What KerasClassifier actually does
    KerasClassifier wraps your Keras model in a special object that:

        * Has .fit(), .predict(), .score() methods that scikit-learn recognizes
        * Internally, when you call .fit() on it, it just calls your Keras model's .fit() behind the scenes
        * So scikit-learn thinks it's talking to a normal scikit-learn model, but really it's just talking to your Keras model through this wrapper

    Simple analogy
    Imagine scikit-learn only accepts USB-C chargers, but your Keras model has a USB-A plug.

    KerasClassifier is like a USB-A → USB-C adapter. It doesn't change your charger (the Keras model) — it just lets it plug into a socket (scikit-learn tools) it normally couldn't fit into directly.

    So that's what "behave like a scikit-learn estimator" means — it's not literally becoming a scikit-learn model, it's just wrapped so scikit-learn tools can use it as if it were one.

Situation	Tool to use

Good question! They're not competing for the same job — think of it as two different tuning philosophies, and once you pick scikit-learn's approach (GridSearchCV), you generally don't need KerasTuner at all. Let me break down why both exist and when each shines.



Why KerasTuner exists (its own advantages)
KerasTuner is built specifically for deep learning, so it has smart features that plain GridSearchCV doesn't:

GridSearchCV (via SciKeras)
   → tries EVERY combination you specify (exhaustive)
   → no early stopping of "bad" trials
   → no smart search strategy
   → can be extremely slow for big search spaces

KerasTuner
   → Hyperband: kills bad trials early (saves huge time)
   → BayesianOptimization: learns from previous trials to guess better ones
   → Built-in support for tuning architecture (number of layers, units, etc.)
   → Integrates directly with Keras training loop (callbacks, epochs, etc.)
So why would anyone use SciKeras + GridSearchCV then?
Because sometimes you're not trying to do heavy-duty neural architecture search — you just want to:

Reuse familiar scikit-learn syntax you already know
Combine your Keras model with a scikit-learn Pipeline (e.g., scaling → PCA → Keras model)
Use cross_val_score for quick, simple cross-validation
Tune a small number of hyperparameters (like batch size, epochs, optimizer) where exhaustive search is fine
The real distinction
SciKeras + GridSearchCV
   → best when Keras model is just ONE STEP in a larger sklearn workflow
   → good for small/simple hyperparameter grids

KerasTuner
   → best when you're SERIOUSLY tuning a deep learning model
   → good for large search spaces, architecture search, efficient search strategies
Do people use both together?
Rarely — usually you pick one path:

Path A (sklearn-first): SciKeras KerasClassifier + GridSearchCV/Pipeline — good if your project is mostly scikit-learn based, and Keras is just "one more model."
Path B (Keras-first): KerasTuner — good if your project is deep-learning-focused and you want efficient, smart tuning of architecture and hyperparameters.
So to directly answer you: KerasTuner isn't made obsolete by SciKeras. It exists because it's a smarter, faster, purpose-built tool for tuning deep learning models — while SciKeras's KerasClassifier exists just to make a Keras model "look like" a scikit-learn model so it can fit into that ecosystem. You choose one based on your workflow, not because one is outdated.

````python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, LabelEncoder, OneHotEncoder
from sklearn.pipeline import Pipeline
from scikeras.wrappers import KerasClassifier
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping
import pickle
````

````python
data=pd.read_csv('Churn_Modelling.csv')
data = data.drop(['RowNumber', 'CustomerId', 'Surname'], axis=1)

label_encoder_gender = LabelEncoder()
data['Gender' ] = label_encoder_gender. fit_transform(data[ 'Gender' ])

onehot_encoder_geo = OneHotEncoder(handle_unknown='ignore',sparse_output=False)
geo_encoded = onehot_encoder_geo.fit_transform(data[['Geography' ]])
geo_encoded_df = pd.DataFrame(geo_encoded, columns=onehot_encoder_geo.get_feature_names_out(['Geography']))

data = pd.concat([data.drop('Geography', axis=1), geo_encoded_df], axis=1)

X = data.drop('Exited', axis=1)
y = data['Exited' ]

train_X, test_X, train_y, test_y = train_test_split(X, y, test_size=0.2, random_state=0)


scaler=StandardScaler()
train_X = scaler.fit_transform(train_X)
test_X = scaler.transform(test_X)

with open("label_encoder_gender_tune.pkl", "wb") as f:
    pickle.dump(label_encoder_gender, f)

with open("onehot_encoder_geo_tune.pkl", "wb") as f:
    pickle.dump(onehot_encoder_geo, f)

with open("scaler_tune.pkl", "wb") as f:
    pickle.dump(scaler, f)
````

````python
# Define a function to create the model and try different hyperparameters(kerasClassifier)

def create_model(neurons=32,layers=1):
    model=Sequential()
    model.add(Dense(neurons, activation='relu', input_shape=(train_X.shape[1],)))

    for _ in range(layers-1):
        model.add(Dense(neurons,activation='relu'))

    model.add(Dense(1,activation='sigmoid'))
    model.compile(optimizer='adam',loss="binary_crossentropy",metrics=['accuracy' ])

    return model
````

````python
## Create a KerasClassifier for use in scikit-learn
model = KerasClassifier(model=create_model, verbose=1, layers=1, neurons=32 )
````

````python
# Define the grid search parameters
param_grid = {
    'neurons': [16, 32, 64],
    'layers': [1, 2],
    'epochs': [50, 100]
}
````

````python
# perform grid search with cross-validation
grid = GridSearchCV(estimator=model, param_grid=param_grid, cv=3, n_jobs=-1)
grid_result = grid.fit(train_X, train_y)

print("Best parameters: ", grid_result.best_params_)
print("Best score: ", grid_result.best_score_)

print("Best: %f using %s" % (grid_result.best_score_, grid_result.best_params_))
````
